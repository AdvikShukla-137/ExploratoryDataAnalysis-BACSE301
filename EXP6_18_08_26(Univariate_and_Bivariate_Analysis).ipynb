```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("WineQT.csv")
df.head(3)
```

```python
df.tail(3)
```

```python
df.dtypes
```

```python
print(df.describe())
pd.isna(df).any()
```

```python
#cleaning
```

```python
df['quality'].value_counts()
```

```python
df['quality'].unique()
```

```python
quality_col = df['quality']
print("Quality unique values:", quality_col.unique())
print("Quality mean:", quality_col.mean())
print("Quality median:", quality_col.median())
print("Quality mode:", quality_col.mode()[0])

quality_col.head(10)
```

```python
alcohol_col = df['alcohol']
print("Alcohol unique values:", alcohol_col.unique())
print("Alcohol mean:", alcohol_col.mean())
print("Alcohol median:", alcohol_col.median())
print("Alcohol mode:", alcohol_col.mode()[0])
alcohol_col.head()
```

```python
pH_col = df['pH']
print("pH unique values:", pH_col.unique())
print("pH mean:", pH_col.mean())
print("pH median:", pH_col.median())
print("pH mode:", pH_col.mode()[0])
pH_col.head()
```

```python
mean_alcohol = df["alcohol"].mean()
median_alcohol = df["alcohol"].median()
mode_alcohol = df["alcohol"].mode()
print("Alcohol Mean:", mean_alcohol)
print("Alcohol Median:", median_alcohol)
print("Alcohol Mode:", mode_alcohol)
```

```python
df['quality'].value_counts().sort_index().plot(kind='bar', figsize=(10,6))
plt.title("Distribution of Wine Quality")
plt.ylabel('Number of Wines')
plt.xlabel('Quality')
plt.xticks(rotation=0)
plt.show()
```

```python
sns.FacetGrid(df).map(sns.histplot, "quality", kde=True).add_legend()
```

```python
sns.FacetGrid(df).map(sns.histplot, "alcohol", kde=True).add_legend()
```

```python
sns.boxplot(x="quality",data=df)
plt.show()
```

```python
#bivariate
```

```python
plt.scatter(df["quality"], df["alcohol"])
plt.title("Scatter Plot for Quality vs Alcohol")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
```

```python
sns.boxplot(x="quality",y="fixed acidity",data=df)
plt.show()
```

```python
sns.boxplot(x="quality", y="volatile acidity",data=df)
```

```python
#multivariate
```

```python
sns.pairplot(df,vars = ['quality', 'alcohol','fixed acidity'], kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df,height=2,vars = ['fixed acidity', 'volatile acidity','citric acid'], hue="quality")
plt.show()
```

```python
from scipy import stats

corr = stats.pearsonr(df["quality"], df["alcohol"])
print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
correlation = df.corr(method='pearson', numeric_only=True)
correlation
```

```python
sns.heatmap(correlation,xticklabels=correlation.columns,
            yticklabels=correlation.columns)
```
