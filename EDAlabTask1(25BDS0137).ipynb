```python
import pandas as pd
import numpy as np
df=pd.read_csv("Student_Data.csv")
```

```python
#To get information like no. of rows , columns, data types,Missing Values.
df.info()
```

```python
#prints out specified columns
df[["Name","Attendance"]]
```

```python
#filters out data based on condition for specified column
df[df["Marks"]>80]
```

```python
#sorts based on marks
df.sort_values(by="Marks")
```

```python
#descending order sorting
df.sort_values(by="Marks",ascending=False)
```

```python
#print the mean for the chosen column
print(df["Marks"].mean())
```

```python
#prints entire row with the condition of marks equal to max marks
df[df["Marks"]==df["Marks"].max()]
```

```python
#prints sum of the entire column
print(df["Marks"].sum())
```

```python
#Number of unique values in the dept column
df["Department"].nunique()
```

```python
#merging two datasets(it findss rows with same value for a column and combines the information from those rows into one row)
df2 = pd.DataFrame({
    'Student_ID': ['25B01','25B02','25B08','25B09','25B15','25B16','25B20'],
    'Project_Progress(%)': [90,50,60,33,75,45,10]
})

pd.merge(df, df2, on="Student_ID")
#Finally only prints the ones it merged
```

```python
df3=pd.DataFrame({
    'Student_ID':('25B21','25B22','25B23'),
    'Name':('James','Richard','Clarkson'),
    'Department':('DESIGN','MECH','CSE'),
    'Marks':('92','88','94'),
    'Attendance':('75','88','99')
})
pd.concat([df,df3])
```
