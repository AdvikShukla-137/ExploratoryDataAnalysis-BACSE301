```python
import pandas as pd
distance = [120, 22, 2125, 271, 121, 523, 137, 237, 61, 145, 41, 12]
bins = [0, 25, 75, 200, 500]
category = pd.cut(distance, bins)
print(category)
```

```python
pd.value_counts(category)
```

```python
bin_names = ['Very Short Drive', 'Short Drive', 'Long Drive', 'Very Long Drive']
pd.cut(distance, bins, labels=bin_names)
pd.value_counts(pd.cut(distance,bins,labels=bin_names))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
#Creating a dummy dataset with an outlier
#here we have the outlier data as 120000 in the salary column
data = {
 'Age': [25, 30, 45, 35, 22],
 'Salary': [50000, 64000, 120000, 85000, 1200000]
}
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")
#Min-Max Scaling(Normalization)
#Rescaling generally from 0 to 1
min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
 min_max_scaler.fit_transform(df),
 columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")
#Z-score Standardization
#we get the mean as 0 and std dev as 1 here
standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
 standard_scaler.fit_transform(df),
 columns=df.columns
)
print("Standardized Data:\n", df_standard, "\n")
```
