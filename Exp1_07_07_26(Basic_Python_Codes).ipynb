```python
#counting number of even and odd numbers
n=int(input("Enter Number of Terms: "))
a=[]
e=0
o=0
for i in range(n):
    x=int(input())
    a.append(x)
for i in a:
    if i%2==0:
        e+=1
    else:
        o+=1
print("No. of Even nums=",e)
print("No. of Odd nums=",o)
```

```python
#bubble sort
n=int(input("Enter number of Terms: "))
a=[]
for i in range(n):
    x=int(input())
    a.append(x)
for i in range(n):
    for j in range(n-i-1):
        if a[j]>a[j+1]:
            a[j], a[j+1]=a[j+1], a[j]
print("Sorted List:",a)
```

```python
#Linear Searching
n=int(input("Enter number of elements: "))
a=[]
for i in range(n):
    x = int(input())
    a.append(x)
x=int(input("Enter element to search: "))
for i in range(n):
    if a[i]==x:
        print("Found at position", i + 1)
        break
else :
    print("Not Found")
```

```python
#Factorial of a number
n=int(input("Enter number: "))
f=1
i=1
while i<=n :
    f=f*i
    i+=1
print("Factorial is=",f)
```

```python
#Fibonacci Series
n=int(input("Enter number of terms: "))
a=0
b=1
for i in range(n):
    print(a,end=" ")
    c=a+b
    a=b
    b=c
```

```python
#Checking if a number is prime
n=int(input("Enter a number: "))
if n<2:
    print("Not Prime")
else:
    for i in range(2,n):
        if n%i==0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
#Reversing a number
n=int(input("Enter number: "))
r=0
while n>0 :
    d = n%10
    r = r*10+d
    n =n//10
print("Reversed Number =",r)
```

```python
#Largest Number
n=int(input("Enter number of terms to compare: "))
a=[]
for i in range(n):
    x=int(input())
    a.append(x)
m=a[0]
for i in a:
    if i>m:
        m=i
print("Largest Number is=",m)
```

```python
#Average of Marks Entered
n=int(input("Enter number of subjects: "))
s=0
for i in range(n):
    x=int(input("Enter marks:"))
    s+=x
print("Total Marks=",s)
print("Average=",s/n)
```

```python
#phonebook using a dict
n=int(input("Enter number of contacts: "))
d={}
for i in range(n):
    name=input("Enter name: ")
    num=input("Enter Phone No.: ")
    d[name]=num
x=input("Enter name to search: ")
if x in d:
    print("Phone No.:",d[x])
else:
    print("Contact not found")
```
